# NumPy Array Slicing

##### Slice elements from index 1 to index 5 from the following array:

```python
arr = np.array([1, 2, 3, 4, 5, 6, 7])
print(arr[1:4])
```

##### Slice elements from index 4 to the end of the array:

```python
print(arr[4:])
```

##### Slice elements from the beginning to index 4 (not included):

```python
print(arr[:4])
```

#### Negative Slicing

```python
print(arr[-3:-1])
```

## STEP

```python
arr = np.array([1, 2, 3, 4, 5, 6, 7])
print(arr[1:5])
print(arr[1:7:2])
```

## NumPy Mini Project for Students
Mini Project: Student Marks Analysis Using NumPy
Objective: Create a Python program using the NumPy library to store, process, and analyze student marks.

Project Description:

A college wants to analyze the examination performance of students in three subjects: Python, Java, and Mathematics.

Develop a Python program using NumPy arrays to perform the following operations.
Dataset
Create a NumPy array containing the marks of 5 students in 3 subjects.
Student	Python	Java	Mathematics
Student 1	85	78	90
Student 2	70	88	75
Student 3	95	92	89
Student 4	60	65	70
Student 5	80	85	88

Questions / Tasks
1. Create a 2D NumPy array to store the marks of 5 students.
2. Display the marks of all students.
3. Display the shape and dimensions of the array.
4. Calculate the total marks obtained by each student.
5. Calculate the average marks of each student.
6. Find the highest mark obtained in the entire dataset.
7. Find the lowest mark obtained in the entire dataset.
8. Find the student who scored the highest total marks.
9. Find the student who scored the lowest total marks.
10. Calculate the average marks for each subject.
11. Display the students whose average marks are greater than 80.
12. Find the number of students who scored above 80 in Python.
13. Sort the students based on their total marks in descending order.
14. Add 5 grace marks to every student's Mathematics mark, ensuring no mark exceeds 100.
15. Display the updated marks array.


```python

import numpy as np

# 1. Create a 2D NumPy array
marks = np.array([
    [85, 78, 90],
    [70, 88, 75],
    [95, 92, 89],
    [60, 65, 70],
    [80, 85, 88]
])

students = np.array([
    "Student 1", "Student 2", "Student 3",
    "Student 4", "Student 5"
])

subjects = ["Python", "Java", "Mathematics"]

# 2. Display marks of all students
print("Student Marks:")
print(marks)

# 3. Display shape and dimensions
print("\nShape:", marks.shape)
print("Dimensions:", marks.ndim)

# 4. Calculate total marks of each student
total = np.sum(marks, axis=1)
print("\nTotal Marks:")
print(total)

# 5. Calculate average marks of each student
average = np.mean(marks, axis=1)
print("\nAverage Marks:")
print(average)

# 6. Find highest mark
print("\nHighest Mark:", np.max(marks))

# 7. Find lowest mark
print("Lowest Mark:", np.min(marks))

# 8. Find highest-scoring student
highest = np.argmax(total)
print("\nHighest Scoring Student:", students[highest])
print("Total Marks:", total[highest])

# 9. Find lowest-scoring student
lowest = np.argmin(total)
print("\nLowest Scoring Student:", students[lowest])
print("Total Marks:", total[lowest])

# 10. Calculate subject-wise average
subject_avg = np.mean(marks, axis=0)
print("\nSubject-wise Average:")
for i in range(3):
    print(subjects[i], ":", round(subject_avg[i], 2))

# 11. Students with average above 80
print("\nStudents with Average Above 80:")
print(students[average > 80])

# 12. Count students scoring above 80 in Python
count = np.sum(marks[:, 0] > 80)
print("\nStudents Scoring Above 80 in Python:", count)

# 13. Sort students based on total marks
order = np.argsort(total)[::-1]

print("\nStudents Sorted by Total Marks:")
for i in order:
    print(students[i], ":", total[i])

# 14. Add 5 grace marks to Mathematics
marks[:, 2] = np.clip(marks[:, 2] + 5, 0, 100)

# 15. Display updated marks
print("\nUpdated Marks After Grace Marks:")
print(marks)
```

## Important NumPy Concepts for Students
- axis=0 performs operations column-wise. Here, it calculates subject-wise averages.
- axis=1 performs operations row-wise. Here, it calculates each student's total and average.
- marks[:, 0] selects all students' Python marks.
- marks[:, 2] selects all students' Mathematics marks.
- np.argmax() returns the index of the highest value.
- np.argsort()[::-1] gives indices sorted in descending order.
- np.clip() restricts values to a specified range.
