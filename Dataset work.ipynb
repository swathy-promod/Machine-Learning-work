## Create a student dataset containing 5 students.


```python
import pandas as pd

data = {
    'Student_ID': [1, 2, 3, 4, 5],
    'Name': ['Anu', 'Rahul', 'Meera', 'Arun', 'Priya'],
    'Age': [20, 21, 19, 22, 20],
    'Gender': ['Female', 'Male', 'Female', 'Male', 'Female'],
    'Height': [160.5, 175.2, 158.7, 180.3, 165.4],
    'Mark': [85, 90, 75, 88, 95]
}

df = pd.DataFrame(data)

print(df)
```

## Identify the rows and columns in the dataset.


```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("Column Names:")
print(df.columns)
```

###### Explanation:
- shape[0] gives the number of rows.
- shape[1] gives the number of columns.
- columns displays all column names.


## 3. Identify the features and label.

```python
features = df[['Student_ID', 'Name', 'Age', 'Gender', 'Height']]

label = df['Mark']

print("Features:")
print(features)

print("\nLabel:")
print(label)
```

##### Explanation:
- df[[...]] selects multiple columns as features.
- df['Mark'] selects the target column.
- Features are the input variables.
- Label is the output variable.
Note: Student_ID and Name are included here only to identify the requested columns. They would generally be excluded when training a prediction model.

## 4. Classify the columns into numerical and categorical data.

```python
numerical = df.select_dtypes(include='number')

categorical = df.select_dtypes(include='object')

print("Numerical Data:")
print(numerical)

print("\nCategorical Data:")
print(categorical)
```

###### Explanation:
- select_dtypes(include='number') selects numerical columns.
- select_dtypes(include='object') selects text-based columns.

Output classification:

Numerical:
- Student_ID
- Age
- Height
- Mark
Categorical:
- Name
- Gender

## 5. Identify which columns contain discrete and continuous data.

```python
discrete = df[['Student_ID', 'Age', 'Mark']]

continuous = df[['Height']]

print("Discrete Data:")
print(discrete)

print("\nContinuous Data:")
print(continuous)
```

###### Explanation:
- Discrete data contains countable values.
- Continuous data contains measurable values that can include decimals.
- df[['column']] selects the specified columns.
Output classification:

Discrete:
- Student_ID
- Age
- Mark
Continuous:

- Height

Note: These classifications are based on how the values are recorded in this example.

## 6. Save the dataset as a CSV file.

```python
df.to_csv('students.csv', index=False)
```

###### Explanation:
- to_csv() saves the DataFrame as a CSV file.
- 'students.csv' is the file name.
- index=False prevents Pandas from saving an additional index column.

## 7. Load the CSV file using Pandas.





```python
import pandas as pd
data = pd.read_csv('students.csv')
print(data)
```

- data stores the loaded dataset.

### Complete Python Program
Here is the complete program combining all the practical exercises.


```python
import pandas as pd
# 1. Create a student dataset
data = {    'Student_ID': [1, 2, 3, 4, 5],    'Name': ['Anu', 'Rahul', 'Meera', 'Arun', 'Priya'],    'Age': [20, 21, 19, 22, 20],    'Gender': ['Female', 'Male', 'Female', 'Male', 'Female'],    'Height': [160.5, 175.2, 158.7, 180.3, 165.4],    'Mark': [85, 90, 75, 88, 95]}
df = pd.DataFrame(data)
# Display datasetprint("Student Dataset:")
print(df)
# 2. Identify rows and columns
print("\nRows and Columns:")
print(df.shape)
# 3. Features and Label
print("\nFeatures:")
print(df[['Student_ID', 'Name', 'Age', 'Gender', 'Height']])
print("\nLabel:")
print(df['Mark'])
# 4. Save dataset as CSV
df.to_csv('students.csv', index=False)
# 5. Load CSV file
data = pd.read_csv('students.csv')
# 6. Display first five rows
print("\nFirst Five Rows:")
print(data.head())
# 7. Display number of rows and columns
print("\nShape of Dataset:")
print(data.shape)
# 8. Display dataset information
print("\nDataset Information:")
data.info()
```
