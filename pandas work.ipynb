## 5. Find the total and average marks of students

## 10. Find and replace missing values in a DataFrame

```python
import pandas as pd

data = {
    'Name': ['Anu', 'Arun', 'Meera', 'Rahul'],
    'Mark': [85, None, 78, None]
}

df = pd.DataFrame(data)

print("Missing Values:")
print(df.isnull())

df = df.fillna(0)

print("After Replacing Missing Values:")
print(df)
```

## Mini Project for Students
Student Performance Analysis Using Pandas

Create a student dataset containing:

Student Name
Roll Number
Department
Attendance
Python Mark
Java Mark
Total Mark
Average Mark
Perform the following operations:

Create a DataFrame.
Display all student details.
Calculate total marks.
Calculate average marks.
Find the highest-scoring student.
Find the lowest-scoring student.
Display students with an average above 80.
Sort students based on total marks.
Calculate department-wise average marks.
Export the final result into a CSV file.
Important Teaching Point: NumPy vs Pandas
Since NumPy and Pandas are both important libraries in Machine Learning, students should understand their differences.


```python

import pandas as pd

# 1. Create Student Dataset
data = {
    'Student Name': ['Anu', 'Arun', 'Meera', 'Rahul', 'Sneha'],
    'Roll Number': [101, 102, 103, 104, 105],
    'Department': ['BCA', 'BSc CS', 'BCA', 'BSc CS', 'BCA'],
    'Attendance': [90, 85, 95, 80, 92],
    'Python Mark': [85, 90, 78, 92, 88],
    'Java Mark': [80, 85, 75, 95, 90]
}

# 2. Create DataFrame
df = pd.DataFrame(data)

# 3. Display All Student Details
print("Student Details:")
print(df)

# 4. Calculate Total Marks
df['Total Mark'] = df['Python Mark'] + df['Java Mark']

print("\nTotal Marks:")
print(df)

# 5. Calculate Average Marks
df['Average Mark'] = df['Total Mark'] / 2

print("\nAverage Marks:")
print(df)

# 6. Find Highest Scoring Student
highest = df.loc[df['Total Mark'].idxmax()]

print("\nHighest Scoring Student:")
print(highest)

# 7. Find Lowest Scoring Student
lowest = df.loc[df['Total Mark'].idxmin()]

print("\nLowest Scoring Student:")
print(lowest)

# 8. Students With Average Above 80
print("\nStudents With Average Above 80:")
print(df[df['Average Mark'] > 80])

# 9. Sort Students Based on Total Marks
sorted_df = df.sort_values(by='Total Mark', ascending=False)

print("\nStudents Sorted By Total Marks:")
print(sorted_df)

# 10. Department-wise Average Marks
dept_avg = df.groupby('Department')['Average Mark'].mean()

print("\nDepartment-wise Average:")
print(dept_avg)

# 11. Export Final Result to CSV
df.to_csv('student_performance.csv', index=False)

print("\nCSV File Created Successfully!")
```
