### Dataset examples

#### 1.student study hours

```python
import pandas as pd

data = {
    "StudyHours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks": [20, 30, 40, 50, 60, 70, 80, 90],
    "Result": [
        "Fail",
        "Fail",
        "Pass",
        "Pass",
        "Pass",
        "Pass",
        "Distinction",
        "Distinction"
    ]
}

df = pd.DataFrame(data)

print(df)
```

##### Alternatively, if you want to create it directly from a table:

```python
import pandas as pd

df = pd.DataFrame([
    [1, 20, "Fail"],
    [2, 30, "Fail"],
    [3, 40, "Pass"],
    [4, 50, "Pass"],
    [5, 60, "Pass"],
    [6, 70, "Pass"],
    [7, 80, "Distinction"],
    [8, 90, "Distinction"]
], columns=["StudyHours", "Marks", "Result"])

print(df)
```
